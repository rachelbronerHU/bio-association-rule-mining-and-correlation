# Investigation 14: Transactions Composition 3D PCA

This notebook computes the local neighborhood composition (ignoring `_CENTER` and `_NEIGHBOR` roles) for **every single transaction (cell)** across FOVs, and performs a 3D Principal Component Analysis (PCA) on these transactions. It allows you to filter and color by specific organs or disease stages.

## 1. Imports & Configuration

Import required libraries and set global constants for data filtering, processing, and visualization.

```python
import pandas as pd
import numpy as np
import os
import sys
import plotly.express as px
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
import warnings

warnings.filterwarnings('ignore')

# Add root to path for utils
sys.path.append(os.path.abspath('../../'))

# --- Configuration ---
RADIUS = 25.0
MIN_CELLS_PER_PATCH = 2

# Visualization & Filtering Constants
ORGANS_TO_INCLUDE = None # e.g. ['Colon', 'Duodenum'] or None for all
COLOR_BY = 'Pathological score' # Options: 'Pathological score', 'Organ', 'Clinical score'

# Due to plotly browser limits with > 700k points in 3D, you can set this < 1.0 to subsample.
PLOT_SAMPLE_FRACTION = 0.1 
```

## 2. Data Loading & Preparation

Load the cell tables and FOV metadata. We also handle Control/Control_S missing labels and normalize the coordinates so distances are in micrometers (um).

```python
def load_and_prep_data():
    print("Loading datasets...")
    cell_table_path = '../../data/MIBIGutCsv/cell_table.csv'
    fov_meta_path = '../../data/MIBIGutCsv/fovs_metadata.csv'
    biopsy_meta_path = '../../data/MIBIGutCsv/biopsy_metadata.csv'

    df_cells = pd.read_csv(cell_table_path)
    df_fovs = pd.read_csv(fov_meta_path)
    df_biopsy = pd.read_csv(biopsy_meta_path)

    # Merge Metadata
    df_fovs = pd.merge(df_fovs, df_biopsy[['Biopsy_ID', 'Pathological score', 'Clinical score', 'Localization']], 
                       left_on='Patient', right_on='Biopsy_ID', how='left')

    # Fix missing scores
    def get_clean_score(row, score_col):
        if pd.notna(row[score_col]):
            return str(row[score_col])
        if str(row['FOV']).startswith('S_'):
            return 'Control_S'
        return 'Control'
        
    df_fovs["Pathological score"] = df_fovs.apply(lambda row: get_clean_score(row, "Pathological score"), axis=1)
    df_fovs["Clinical score"] = df_fovs.apply(lambda row: get_clean_score(row, "Clinical score"), axis=1)

    # Standardized Organ Enrichment
    def get_organ(row):
        if pd.notna(row.get("Localization")): return row["Localization"]
        cohort = str(row.get("Cohort", ""))
        if "Colon" in cohort: return "Colon"
        if "Duodenum" in cohort: return "Duodenum"
        return "Unknown"

    df_fovs["Organ"] = df_fovs.apply(get_organ, axis=1)
    
    if ORGANS_TO_INCLUDE is not None:
        df_fovs = df_fovs[df_fovs['Organ'].isin(ORGANS_TO_INCLUDE)]

    # Vectorized Coordinate Normalization
    fov_to_size = df_fovs.set_index('FOV')['Size [um]'].to_dict()
    sizes = df_cells['fov'].map(fov_to_size).fillna(800)
    res = np.where(sizes == 400, 1024, 2048)
    scale = sizes / res
    
    df_cells['x_um'] = df_cells['centroid_x'] * scale
    df_cells['y_um'] = df_cells['centroid_y'] * scale
    
    # Filter cells by valid FOVs
    df_cells = df_cells[df_cells['fov'].isin(df_fovs['FOV'])]

    print(f"Loaded {len(df_cells)} cells across {len(df_fovs)} FOVs.")
    return df_cells, df_fovs

df_cells, df_fovs = load_and_prep_data()
```

## 3. Build Transaction Matrix

For each FOV, we find the local neighborhood for *every* cell. We count the cell types present in that specific cell's neighborhood (the transaction), and record it as an individual row.

```python
def build_all_transactions_matrix(df_cells, df_fovs):
    unique_cell_types = df_cells['cell type'].unique()
    all_transactions = []

    print("Extracting composition for every single transaction (this may take a couple of minutes)...")
    
    # For performance, map cell types to indices
    ctype_to_idx = {ctype: i for i, ctype in enumerate(unique_cell_types)}
    num_types = len(unique_cell_types)
    
    for fov in df_fovs['FOV'].unique():
        fov_data = df_cells[df_cells['fov'] == fov]
        if fov_data.empty:
            continue
            
        coords = fov_data[['x_um', 'y_um']].values
        types = fov_data['cell type'].values
        type_indices = np.array([ctype_to_idx[t] for t in types])
        
        if len(coords) < MIN_CELLS_PER_PATCH:
            continue
            
        nn = NearestNeighbors(radius=RADIUS).fit(coords)
        neighbors_idx = nn.radius_neighbors(coords, return_distance=False)
        
        # Get metadata for the FOV to append to each transaction
        fov_meta = df_fovs[df_fovs['FOV'] == fov].iloc[0]
        organ = fov_meta['Organ']
        stage = fov_meta['Pathological score']
        c_score = fov_meta['Clinical score']
        
        for i, idxs in enumerate(neighbors_idx):
            if len(idxs) < MIN_CELLS_PER_PATCH:
                continue
                
            # Fast counting using bincount
            neighbor_types = type_indices[idxs]
            counts = np.bincount(neighbor_types, minlength=num_types)
            
            # Create row
            row = {
                'FOV': fov,
                'Organ': organ,
                'Pathological score': stage,
                'Clinical score': c_score,
                'Center Cell Type': types[i]
            }
            for ctype_idx, count in enumerate(counts):
                row[unique_cell_types[ctype_idx]] = count
                
            all_transactions.append(row)

    df_matrix = pd.DataFrame(all_transactions)
    print(f"Transaction matrix generated with shape: {df_matrix.shape}")
    return df_matrix, list(unique_cell_types)

df_transactions, feature_cell_types = build_all_transactions_matrix(df_cells, df_fovs)
```

## 4. 3D PCA Visualization

Scale the transaction compositions, compute the top 3 principal components, and plot every transaction in 3D.

```python
def plot_transactions_pca(df_matrix, features):
    # Filter for features actually present
    feature_cols = [c for c in features if c in df_matrix.columns]
    X = df_matrix[feature_cols].fillna(0)
    
    # Scale features
    print("Scaling features and running PCA...")
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    # Run PCA
    pca = PCA(n_components=3)
    components = pca.fit_transform(X_scaled)
    
    # Add PCs to dataframe
    df_plot = df_matrix.copy()
    df_plot['PC1'] = components[:, 0]
    df_plot['PC2'] = components[:, 1]
    df_plot['PC3'] = components[:, 2]
    
    var_explained = pca.explained_variance_ratio_
    print(f"Explained Variance -> PC1: {var_explained[0]:.1%} | PC2: {var_explained[1]:.1%} | PC3: {var_explained[2]:.1%}")
    
    # Subsample if necessary (browser can crash with 700k 3D points)
    if PLOT_SAMPLE_FRACTION < 1.0:
        num_samples = int(len(df_plot) * PLOT_SAMPLE_FRACTION)
        print(f"Subsampling plot to {num_samples} points ({PLOT_SAMPLE_FRACTION*100}%) for browser performance...")
        df_plot = df_plot.sample(n=num_samples, random_state=42)
        
    # Map common colors. You can add more mapping if COLOR_BY is set to something custom
    color_map = {
        'Control': '#2ca02c',   # Green
        'Control_S': '#1f77b4', # Blue
        'Mild': '#ff7f0e',      # Orange
        'Severe': '#d62728'     # Red
    }
        
    # Generate Plotly figure
    print("Generating 3D plot...")
    fig = px.scatter_3d(
        df_plot,
        x='PC1', y='PC2', z='PC3',
        color=COLOR_BY,
        hover_name='Center Cell Type',
        hover_data=['FOV', 'Organ', 'Pathological score'],
        title=f'3D PCA: All Individual Transactions (Colored by {COLOR_BY})',
        labels={
            'PC1': f'PC1 ({var_explained[0]:.1%})',
            'PC2': f'PC2 ({var_explained[1]:.1%})',
            'PC3': f'PC3 ({var_explained[2]:.1%})'
        },
        color_discrete_map=color_map,
        opacity=0.6 # Add some transparency since there are many points
    )
    
    # Optimization for large scatter plots in Plotly
    fig.update_traces(marker=dict(size=3, line=dict(width=0)))
    
    # Cleanup layout
    fig.update_layout(
        scene=dict(
            xaxis_title='PC1',
            yaxis_title='PC2',
            zaxis_title='PC3'
        ),
        margin=dict(l=0, r=0, b=0, t=40)
    )
    fig.show()

plot_transactions_pca(df_transactions, feature_cell_types)
```
